```python
import pandas as pd

# 讀你的 Excel 寬表
wide_path = "NBA 2026 Schedule.xlsx"
df = pd.read_excel(wide_path)

# 前四欄是基本資訊，後面全部都是日期欄位
meta_cols = ["Team", "Games", "Home", "Away"]
date_cols = [c for c in df.columns if c not in meta_cols]

rows = []

for _, row in df.iterrows():
    team = row["Team"]

    for date_col in date_cols:
        opponent = str(row[date_col]).strip()

        # 空白就跳過（那天沒比賽）
        if opponent == "" or opponent.lower() == "nan":
            continue

        # 判斷主客場：有 @ 就是客場
        if opponent.startswith("@"):
            opp_team = opponent[1:]  # 去掉 @
            home_away = "away"
        else:
            opp_team = opponent
            home_away = "home"

        rows.append({
            "date": date_col,
            "team": team,
            "opponent": opp_team,
            "home_away": home_away
        })

# 轉成長表 DataFrame
long_df = pd.DataFrame(rows)

# 排序一下（可選）
long_df = long_df.sort_values(["date", "team"]).reset_index(drop=True)

# 存成你要的 schedule.csv
output_path = "schedule.csv"
long_df.to_csv(output_path, index=False)

print(f"Done! Saved {len(long_df)} rows to {output_path}")
```
