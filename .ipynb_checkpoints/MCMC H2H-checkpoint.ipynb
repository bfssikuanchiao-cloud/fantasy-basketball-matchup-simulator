```python
import pandas as pd
import numpy as np
from nba_api.stats.static import players
from nba_api.stats.endpoints import playergamelog

# ============================================================
# 1. 抓球員逐場資料（保留完整 MATCHUP）
# ============================================================

def get_player_df_from_api(name, season="2025-26"):
    result = players.find_players_by_full_name(name)
    if not result:
        raise ValueError(f"找不到球員：{name}")
    player_id = result[0]["id"]

    df = playergamelog.PlayerGameLog(
        player_id=player_id,
        season=season,
        season_type_all_star="Regular Season"
    ).get_data_frames()[0]

    df = df.rename(columns={
        "GAME_DATE": "Date",
        "MATCHUP": "Matchup",
        "WL": "Result",
        "MIN": "Min",
        "FGM": "FGM",
        "FGA": "FGA",
        "FG_PCT": "FG%",
        "FG3M": "3PM",
        "FG3A": "3PA",
        "FG3_PCT": "3P%",
        "FTM": "FTM",
        "FTA": "FTA",
        "FT_PCT": "FT%",
        "OREB": "OR",
        "DREB": "DR",
        "REB": "Reb",
        "AST": "Ast",
        "STL": "Stl",
        "BLK": "Blk",
        "TOV": "TO",
        "PF": "PF",
        "PLUS_MINUS": "+/-",
        "PTS": "Pts"
    })

    df["Date"] = pd.to_datetime(df["Date"])

    def parse_matchup(m):
        parts = m.split(" ")
        team = parts[0]
        symbol = parts[1]  # vs / @
        opp = parts[2]
        is_home = (symbol == "vs")
        return pd.Series({
            "Team": team,
            "Opp": opp,
            "HomeAway": symbol,
            "IsHome": is_home
        })

    extra = df["Matchup"].apply(parse_matchup)
    df = pd.concat([df, extra], axis=1)

    return df


# ============================================================
# 2. schedule.csv → 賽程與剩餘比賽
# ============================================================

def load_schedule(path="schedule.csv"):
    df = pd.read_csv(path)
    df["date"] = pd.to_datetime(df["date"])
    return df

def get_games_in_range(schedule_df, team, start_date, end_date):
    start_date = pd.to_datetime(start_date)
    end_date   = pd.to_datetime(end_date)
    mask = (
        (schedule_df["team"] == team) &
        (schedule_df["date"] >= start_date) &
        (schedule_df["date"] <= end_date)
    )
    return schedule_df[mask].sort_values("date")


# ============================================================
# 3. 單場加權（decay + 主客場 + 對手）
# ============================================================

def apply_single_game_weight(df, game,
                             decay=0.99,
                             home_boost=1.3,
                             away_boost=1.3,
                             opp_boost=1.8):

    df = df.copy()
    df = df.sort_values("Date", ascending=False).reset_index(drop=True)
    df["Weight"] = decay ** np.arange(len(df))

    next_is_home = (game["home_away"] == "home")
    next_opp = game["opponent"]

    is_home_sample = df["HomeAway"] == "vs"
    is_away_sample = df["HomeAway"] == "@"

    if next_is_home:
        df.loc[is_home_sample, "Weight"] *= home_boost
    else:
        df.loc[is_away_sample, "Weight"] *= away_boost

    df.loc[df["Opp"] == next_opp, "Weight"] *= opp_boost

    df["Weight"] = df["Weight"] / df["Weight"].sum()

    return df


# ============================================================
# 4. 逐場 Monte‑Carlo 抽樣（每一場依 schedule 加權）
# ============================================================

def simulate_future_games(df_player, games_df):
    future_totals = {
        "Pts": 0, "Reb": 0, "Ast": 0, "Stl": 0, "Blk": 0,
        "3PM": 0, "TO": 0, "FGM": 0, "FGA": 0, "FTM": 0, "FTA": 0
    }

    for _, game in games_df.iterrows():

        weighted_df = apply_single_game_weight(df_player, game)

        # ⭐⭐⭐ 加權檢查（最重要） ⭐⭐⭐
        #print("\n=== 加權檢查：", game["date"], game["opponent"], game["home_away"], "===")
        #print(weighted_df[["Date", "Matchup", "HomeAway", "Opp", "Weight"]].head(15))

        sample = weighted_df.sample(n=1, replace=True, weights="Weight")

        for cat in future_totals.keys():
            future_totals[cat] += sample[cat].iloc[0]

    return future_totals


```

```python
# ======== 你輸入的日期區間 ========
start_date = "2026-10-19"
end_date   = "2026-10-25"

# ======== 抓 schedule ========
schedule_df = load_schedule("schedule.csv")

# ======== 抓兩個球員資料 ========
MyPlayer = get_player_df_from_api("Anthony Edwards")
Opponent = get_player_df_from_api("Cason Wallace")

# ======== 找兩隊在日期區間內的所有比賽 ========
my_team = MyPlayer["Team"].iloc[0]
opp_team = Opponent["Team"].iloc[0]

my_games_df = get_games_in_range(schedule_df, my_team, start_date, end_date)
opp_games_df = get_games_in_range(schedule_df, opp_team, start_date, end_date)

print("Dyson 賽程：\n", my_games_df)
print("\nCason 賽程：\n", opp_games_df)

my_games_left = len(my_games_df)
opp_games_left = len(opp_games_df)

print("\nDyson 剩餘比賽數：", my_games_left)
print("Cason 剩餘比賽數：", opp_games_left)


# ======== 目前比分 ========
current_my = {
    "Pts": 120, "Reb": 48, "Ast": 35, "Stl": 24, "Blk": 8,
    "3PM": 18, "TO": 16, "FGM": 50, "FGA": 100, "FTM": 20, "FTA": 25
}

current_opp = {
    "Pts": 118, "Reb": 52, "Ast": 38, "Stl": 26, "Blk": 10,
    "3PM": 17, "TO": 18, "FGM": 48, "FGA": 95, "FTM": 18, "FTA": 22
}


# ============================================================
# 逐場 Monte‑Carlo（每一場依 schedule 加權）
# ============================================================

categories = ["Pts", "Reb", "Ast", "Stl", "Blk", "3PM", "TO", "FG%", "FT%"]

n_sim = 3000
win_counts = {cat: 0 for cat in categories}
margin_dict = {cat: [] for cat in categories}
vector_results = []

for sim in range(n_sim):

    # MyPlayer逐場模擬
    my_future = simulate_future_games(MyPlayer, my_games_df)

    # Opponent 逐場模擬
    opp_future = simulate_future_games(Opponent, opp_games_df)

    # FG / FT 計算
    my_FGM_total = current_my["FGM"] + my_future["FGM"]
    my_FGA_total = current_my["FGA"] + my_future["FGA"]
    my_FTM_total = current_my["FTM"] + my_future["FTM"]
    my_FTA_total = current_my["FTA"] + my_future["FTA"]

    opp_FGM_total = current_opp["FGM"] + opp_future["FGM"]
    opp_FGA_total = current_opp["FGA"] + opp_future["FGA"]
    opp_FTM_total = current_opp["FTM"] + opp_future["FTM"]
    opp_FTA_total = current_opp["FTA"] + opp_future["FTA"]

    my_FG_pct = my_FGM_total / my_FGA_total if my_FGA_total > 0 else 0
    opp_FG_pct = opp_FGM_total / opp_FGA_total if opp_FGA_total > 0 else 0

    my_FT_pct = my_FTM_total / my_FTA_total if my_FTA_total > 0 else 0
    opp_FT_pct = opp_FTM_total / opp_FTA_total if opp_FTA_total > 0 else 0

    win_vector = []

    # FG%
    margin_dict["FG%"].append(my_FG_pct - opp_FG_pct)
    win_vector.append(1 if my_FG_pct > opp_FG_pct else 0)
    win_counts["FG%"] += win_vector[-1]

    # FT%
    margin_dict["FT%"].append(my_FT_pct - opp_FT_pct)
    win_vector.append(1 if my_FT_pct > opp_FT_pct else 0)
    win_counts["FT%"] += win_vector[-1]

    # 其他 7 項
    for cat in ["Pts", "Reb", "Ast", "Stl", "Blk", "3PM", "TO"]:
        my_total = current_my[cat] + my_future[cat]
        opp_total = current_opp[cat] + opp_future[cat]

        margin = my_total - opp_total
        margin_dict[cat].append(margin)

        if cat == "TO":
            win = 1 if my_total < opp_total else 0
        else:
            win = 1 if my_total > opp_total else 0

        win_counts[cat] += win
        win_vector.append(win)

    vector_results.append(win_vector)
```

```python
from collections import Counter
import seaborn as sns
import matplotlib.pyplot as plt

# 單項結果
print("\n" + "=" * 60)
print("CATEGORY WIN PROBABILITIES")
print("=" * 60)

for cat in categories:
    data = margin_dict[cat]
    win_prob = win_counts[cat] / n_sim
    mean_margin = np.mean(data)
    p5, p50, p95 = np.quantile(data, [0.05, 0.50, 0.95])

    print()
    print(cat)
    print(f"Win Prob : {win_prob:.1%}")
    print(f"Avg Margin: {mean_margin:.3f}")
    print(f"P5/P50/P95: {p5:.3f}, {p50:.3f}, {p95:.3f}")

# 0~9 分布
win_counts_total = [sum(v) for v in vector_results]
counter = Counter(win_counts_total)

print("\n" + "=" * 60)
print("OVERALL CATEGORY WIN DISTRIBUTION (0~9)")
print("=" * 60)

for k in range(10):
    prob = counter[k] / n_sim
    print(f"{k}:{9-k} → {prob:.2%}")

# Margin 分布圖
plt.figure(figsize=(10, 9))
for i, cat in enumerate(categories, 1):
    plt.subplot(3, 3, i)
    sns.histplot(margin_dict[cat], bins=20, kde=True)
    plt.axvline(x=0, color="red", linestyle="--")
    plt.title(f"{cat} Margin Distribution")
plt.tight_layout()
plt.show()

# 熱圖
df_heat = pd.DataFrame(vector_results, columns=categories)
win_probs = df_heat.mean() * 100
win_probs = win_probs.round(1)

score_labels = [f"{k}:{9-k}" for k in range(10)]
score_probs = [(counter[k] / n_sim) * 100 for k in range(10)]
df_score = pd.DataFrame([score_probs], columns=score_labels)

fig, axes = plt.subplots(2, 1, figsize=(6, 3))

sns.heatmap(win_probs.to_frame().T,
            annot=True, fmt=".1f",
            cmap="coolwarm", vmin=0, vmax=100,
            ax=axes[0])
axes[0].set_title("Category Win Probability Heatmap (%)")
axes[0].set_xlabel("")
axes[0].set_ylabel("")

sns.heatmap(df_score,
            annot=True, fmt=".2f",
            cmap="coolwarm", vmin=0, vmax=max(score_probs),
            ax=axes[1])
axes[1].set_title("Score Distribution Heatmap (Win : Loss %)")
axes[1].set_xlabel("Win : Loss")
axes[1].set_ylabel("")
axes[1].set_yticks([])

plt.tight_layout()
plt.show()
```
