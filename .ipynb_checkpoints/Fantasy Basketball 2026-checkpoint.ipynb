```python
# ============================================================
# Fantasy Basketball H2H 9-Cat
# N-PLAYER MONTE CARLO SIMULATOR
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from collections import Counter
from nba_api.stats.static import players
from nba_api.stats.endpoints import playergamelog


# ============================================================
# 1. 抓單一球員 NBA 逐場資料
# ============================================================

def get_player_df_from_api(name, season="2025-26"):

    result = players.find_players_by_full_name(name)

    if not result:
        raise ValueError(f"找不到球員：{name}")

    player_id = result[0]["id"]

    df = playergamelog.PlayerGameLog(
        player_id=player_id,
        season=season,
        season_type_all_star="Regular Season"
    ).get_data_frames()[0]

    df = df.rename(columns={
        "GAME_DATE": "Date",
        "MATCHUP": "Matchup",
        "WL": "Result",
        "MIN": "Min",

        "FGM": "FGM",
        "FGA": "FGA",
        "FG_PCT": "FG%",

        "FG3M": "3PM",
        "FG3A": "3PA",
        "FG3_PCT": "3P%",

        "FTM": "FTM",
        "FTA": "FTA",
        "FT_PCT": "FT%",

        "OREB": "OR",
        "DREB": "DR",
        "REB": "Reb",

        "AST": "Ast",
        "STL": "Stl",
        "BLK": "Blk",
        "TOV": "TO",

        "PF": "PF",
        "PLUS_MINUS": "+/-",
        "PTS": "Pts"
    })

    df["Date"] = pd.to_datetime(df["Date"])

    # --------------------------------------------------------
    # 解析歷史比賽對手 + 主客場
    # --------------------------------------------------------

    def parse_matchup(m):

        parts = m.split(" ")

        team = parts[0]
        symbol = parts[1]       # vs / @
        opp = parts[2]

        return pd.Series({
            "Team": team,
            "Opp": opp,
            "HomeAway": symbol,
            "IsHome": symbol == "vs"
        })

    extra = df["Matchup"].apply(parse_matchup)

    df = pd.concat(
        [df, extra],
        axis=1
    )

    return df


# ============================================================
# 2. 建立多人 Roster
# ============================================================

def build_roster(
    player_names,
    season="2025-26"
):

    roster = {}

    for name in player_names:

        print(f"Loading {name} ...")

        roster[name] = get_player_df_from_api(
            name,
            season=season
        )

    return roster


# ============================================================
# 3. 讀取 NBA Schedule
# ============================================================

def load_schedule(path="schedule.csv"):

    df = pd.read_csv(path)

    df["date"] = pd.to_datetime(
        df["date"]
    )

    return df


# ============================================================
# 4. 找某球隊在日期區間內的比賽
# ============================================================

def get_games_in_range(
    schedule_df,
    team,
    start_date,
    end_date
):

    start_date = pd.to_datetime(
        start_date
    )

    end_date = pd.to_datetime(
        end_date
    )

    mask = (
        (schedule_df["team"] == team)
        &
        (schedule_df["date"] >= start_date)
        &
        (schedule_df["date"] <= end_date)
    )

    return (
        schedule_df[mask]
        .sort_values("date")
    )


# ============================================================
# 5. 單場 Historical Game Weight
#
# Weight =
# Recency
# × Home/Away
# × Opponent
# ============================================================

def apply_single_game_weight(
    df,
    game,
    decay=0.99,
    home_boost=1.3,
    away_boost=1.3,
    opp_boost=1.8
):

    df = df.copy()

    # 最新 historical game 排第一
    df = (
        df
        .sort_values(
            "Date",
            ascending=False
        )
        .reset_index(drop=True)
    )

    # --------------------------------------------------------
    # Recency Weight
    # --------------------------------------------------------

    df["Weight"] = (
        decay **
        np.arange(len(df))
    )

    # --------------------------------------------------------
    # Future game environment
    # --------------------------------------------------------

    next_is_home = (
        game["home_away"] == "home"
    )

    next_opp = game["opponent"]

    # --------------------------------------------------------
    # Historical game environment
    # --------------------------------------------------------

    is_home_sample = (
        df["HomeAway"] == "vs"
    )

    is_away_sample = (
        df["HomeAway"] == "@"
    )

    # --------------------------------------------------------
    # Home / Away Boost
    # --------------------------------------------------------

    if next_is_home:

        df.loc[
            is_home_sample,
            "Weight"
        ] *= home_boost

    else:

        df.loc[
            is_away_sample,
            "Weight"
        ] *= away_boost

    # --------------------------------------------------------
    # Same Opponent Boost
    # --------------------------------------------------------

    df.loc[
        df["Opp"] == next_opp,
        "Weight"
    ] *= opp_boost

    # --------------------------------------------------------
    # Normalize → sampling probabilities
    # --------------------------------------------------------

    df["Weight"] = (
        df["Weight"]
        /
        df["Weight"].sum()
    )

    return df


# ============================================================
# 6. 模擬「單一球員」剩餘所有比賽
# ============================================================

def simulate_future_games(
    df_player,
    games_df
):

    future_totals = {

        "Pts": 0,
        "Reb": 0,
        "Ast": 0,
        "Stl": 0,
        "Blk": 0,
        "3PM": 0,
        "TO": 0,

        "FGM": 0,
        "FGA": 0,

        "FTM": 0,
        "FTA": 0
    }

    # 每一場 future NBA game
    for _, game in games_df.iterrows():

        # 依照這一場的
        # opponent + home/away
        # 重新建立 historical weights

        weighted_df = apply_single_game_weight(
            df_player,
            game
        )

        # ----------------------------------------------------
        # JOINT EMPIRICAL BOOTSTRAP
        #
        # 抽的是一整場 Box Score
        # 不是每個 Category 獨立抽
        # ----------------------------------------------------

        sample = weighted_df.sample(
            n=1,
            replace=True,
            weights="Weight"
        )

        for stat in future_totals:

            future_totals[stat] += (
                sample[stat].iloc[0]
            )

    return future_totals


# ============================================================
# 7. 模擬「整支 Fantasy Roster」
# ============================================================

def simulate_roster(
    roster,
    schedule_df,
    start_date,
    end_date
):

    roster_totals = {

        "Pts": 0,
        "Reb": 0,
        "Ast": 0,
        "Stl": 0,
        "Blk": 0,
        "3PM": 0,
        "TO": 0,

        "FGM": 0,
        "FGA": 0,

        "FTM": 0,
        "FTA": 0
    }

    # --------------------------------------------------------
    # 每一位球員
    # --------------------------------------------------------

    for player_name, player_df in roster.items():

        team = (
            player_df["Team"]
            .iloc[0]
        )

        # 該球員所屬 NBA Team
        # 在指定期間的所有比賽

        games_df = get_games_in_range(
            schedule_df,
            team,
            start_date,
            end_date
        )

        # 模擬該球員所有 future games

        player_future = simulate_future_games(
            player_df,
            games_df
        )

        # 加入 Fantasy Team totals

        for stat in roster_totals:

            roster_totals[stat] += (
                player_future[stat]
            )

    return roster_totals


# ============================================================
# 8. 使用者設定
# ============================================================

start_date = "2026-10-19"
end_date   = "2026-10-25"

season = "2025-26"

n_sim = 3000


# ============================================================
# 9. 輸入雙方 Roster
#
# 目前先手動輸入。
# 未來 Yahoo API 會取代這一段。
# ============================================================

my_player_names = [

    "Anthony Edwards",
    "Dyson Daniels",
    "Nikola Jokic"

]


opp_player_names = [

    "Cason Wallace",
    "Jalen Williams",
    "Chet Holmgren"

]


# ============================================================
# 10. NBA Schedule
# ============================================================

schedule_df = load_schedule(
    "schedule.csv"
)


# ============================================================
# 11. 自動抓取所有球員歷史資料
# ============================================================

print("\nLoading MY roster")
print("=" * 60)

my_roster = build_roster(
    my_player_names,
    season=season
)


print("\nLoading OPPONENT roster")
print("=" * 60)

opp_roster = build_roster(
    opp_player_names,
    season=season
)


# ============================================================
# 12. 顯示各球員剩餘比賽數
# ============================================================

print("\nMY TEAM SCHEDULE")
print("=" * 60)

for name, df in my_roster.items():

    team = df["Team"].iloc[0]

    games = get_games_in_range(
        schedule_df,
        team,
        start_date,
        end_date
    )

    print(
        f"{name:25} "
        f"{team:4} "
        f"{len(games)} games"
    )


print("\nOPPONENT TEAM SCHEDULE")
print("=" * 60)

for name, df in opp_roster.items():

    team = df["Team"].iloc[0]

    games = get_games_in_range(
        schedule_df,
        team,
        start_date,
        end_date
    )

    print(
        f"{name:25} "
        f"{team:4} "
        f"{len(games)} games"
    )


# ============================================================
# 13. 目前 Fantasy Matchup Totals
#
# 目前手動輸入
# 未來 Yahoo API 取代
# ============================================================

current_my = {

    "Pts": 120,
    "Reb": 48,
    "Ast": 35,
    "Stl": 24,
    "Blk": 8,
    "3PM": 18,
    "TO": 16,

    "FGM": 50,
    "FGA": 100,

    "FTM": 20,
    "FTA": 25
}


current_opp = {

    "Pts": 118,
    "Reb": 52,
    "Ast": 38,
    "Stl": 26,
    "Blk": 10,
    "3PM": 17,
    "TO": 18,

    "FGM": 48,
    "FGA": 95,

    "FTM": 18,
    "FTA": 23
}


# ============================================================
# 14. 9 Categories
# ============================================================

categories = [

    "FG%",
    "FT%",

    "Pts",
    "Reb",
    "Ast",
    "Stl",
    "Blk",
    "3PM",
    "TO"
]


# ============================================================
# 15. Monte Carlo storage
# ============================================================

win_counts = {
    cat: 0
    for cat in categories
}

margin_dict = {
    cat: []
    for cat in categories
}

vector_results = []


# ============================================================
# 16. MONTE CARLO
# ============================================================

for sim in range(n_sim):

    # --------------------------------------------------------
    # 模擬整支球隊
    # --------------------------------------------------------

    my_future = simulate_roster(
        my_roster,
        schedule_df,
        start_date,
        end_date
    )

    opp_future = simulate_roster(
        opp_roster,
        schedule_df,
        start_date,
        end_date
    )


    # ========================================================
    # FG%
    # ========================================================

    my_FGM_total = (
        current_my["FGM"]
        + my_future["FGM"]
    )

    my_FGA_total = (
        current_my["FGA"]
        + my_future["FGA"]
    )


    opp_FGM_total = (
        current_opp["FGM"]
        + opp_future["FGM"]
    )

    opp_FGA_total = (
        current_opp["FGA"]
        + opp_future["FGA"]
    )


    my_FG_pct = (
        my_FGM_total / my_FGA_total
        if my_FGA_total > 0
        else 0
    )

    opp_FG_pct = (
        opp_FGM_total / opp_FGA_total
        if opp_FGA_total > 0
        else 0
    )


    # ========================================================
    # FT%
    # ========================================================

    my_FTM_total = (
        current_my["FTM"]
        + my_future["FTM"]
    )

    my_FTA_total = (
        current_my["FTA"]
        + my_future["FTA"]
    )


    opp_FTM_total = (
        current_opp["FTM"]
        + opp_future["FTM"]
    )

    opp_FTA_total = (
        current_opp["FTA"]
        + opp_future["FTA"]
    )


    my_FT_pct = (
        my_FTM_total / my_FTA_total
        if my_FTA_total > 0
        else 0
    )

    opp_FT_pct = (
        opp_FTM_total / opp_FTA_total
        if opp_FTA_total > 0
        else 0
    )


    # ========================================================
    # 本次 simulation 的 9 Cat win/loss
    # ========================================================

    win_vector = []


    # ========================================================
    # FG%
    # ========================================================

    fg_margin = (
        my_FG_pct
        - opp_FG_pct
    )

    margin_dict["FG%"].append(
        fg_margin
    )

    fg_win = int(
        my_FG_pct > opp_FG_pct
    )

    win_counts["FG%"] += fg_win

    win_vector.append(
        fg_win
    )


    # ========================================================
    # FT%
    # ========================================================

    ft_margin = (
        my_FT_pct
        - opp_FT_pct
    )

    margin_dict["FT%"].append(
        ft_margin
    )

    ft_win = int(
        my_FT_pct > opp_FT_pct
    )

    win_counts["FT%"] += ft_win

    win_vector.append(
        ft_win
    )


    # ========================================================
    # Counting Stats
    # ========================================================

    for cat in [

        "Pts",
        "Reb",
        "Ast",
        "Stl",
        "Blk",
        "3PM",
        "TO"

    ]:

        my_total = (
            current_my[cat]
            + my_future[cat]
        )

        opp_total = (
            current_opp[cat]
            + opp_future[cat]
        )


        margin = (
            my_total
            - opp_total
        )

        margin_dict[cat].append(
            margin
        )


        # TO 越少越好

        if cat == "TO":

            win = int(
                my_total < opp_total
            )

        else:

            win = int(
                my_total > opp_total
            )


        win_counts[cat] += win

        win_vector.append(
            win
        )


    # --------------------------------------------------------
    # 保存這個 Monte Carlo world
    # --------------------------------------------------------

    vector_results.append(
        win_vector
    )


# ============================================================
# 17. Category Results
# ============================================================

print("\n")
print("=" * 60)
print("CATEGORY WIN PROBABILITIES")
print("=" * 60)


for cat in categories:

    data = margin_dict[cat]

    win_prob = (
        win_counts[cat]
        / n_sim
    )

    mean_margin = np.mean(
        data
    )

    p5, p50, p95 = np.quantile(
        data,
        [0.05, 0.50, 0.95]
    )


    print()

    print(cat)

    print(
        f"Win Prob   : "
        f"{win_prob:.1%}"
    )

    print(
        f"Avg Margin : "
        f"{mean_margin:.3f}"
    )

    print(
        f"P5/P50/P95 : "
        f"{p5:.3f}, "
        f"{p50:.3f}, "
        f"{p95:.3f}"
    )


# ============================================================
# 18. Final Score Distribution
# ============================================================

win_counts_total = [

    sum(v)
    for v in vector_results

]


counter = Counter(
    win_counts_total
)


print("\n")
print("=" * 60)
print("OVERALL CATEGORY WIN DISTRIBUTION")
print("=" * 60)


for k in range(10):

    prob = (
        counter[k]
        / n_sim
    )

    print(
        f"{k}:{9-k} "
        f"→ {prob:.2%}"
    )


# ============================================================
# 19. Strategic Probabilities
# ============================================================

wins_array = np.array(
    win_counts_total
)


p_ge_5 = np.mean(
    wins_array >= 5
)

p_ge_6 = np.mean(
    wins_array >= 6
)

p_ge_7 = np.mean(
    wins_array >= 7
)


print("\n")
print("=" * 60)
print("MATCHUP TARGET PROBABILITIES")
print("=" * 60)

print(
    f"P(至少 5-4) = "
    f"{p_ge_5:.1%}"
)

print(
    f"P(至少 6-3) = "
    f"{p_ge_6:.1%}"
)

print(
    f"P(至少 7-2) = "
    f"{p_ge_7:.1%}"
)


# ============================================================
# 20. Margin Distribution Plots
# ============================================================

plt.figure(
    figsize=(12, 10)
)


for i, cat in enumerate(
    categories,
    1
):

    plt.subplot(
        3,
        3,
        i
    )

    sns.histplot(
        margin_dict[cat],
        bins=20,
        kde=True
    )

    plt.axvline(
        x=0,
        color="red",
        linestyle="--"
    )

    plt.title(
        f"{cat} Margin Distribution"
    )


plt.tight_layout()

plt.show()


# ============================================================
# 21. Heatmaps
# ============================================================

df_heat = pd.DataFrame(
    vector_results,
    columns=categories
)


win_probs = (
    df_heat.mean()
    * 100
).round(1)


score_labels = [

    f"{k}:{9-k}"
    for k in range(10)

]


score_probs = [

    (
        counter[k]
        / n_sim
    ) * 100

    for k in range(10)

]


df_score = pd.DataFrame(
    [score_probs],
    columns=score_labels
)


fig, axes = plt.subplots(
    2,
    1,
    figsize=(9, 4)
)


sns.heatmap(
    win_probs.to_frame().T,
    annot=True,
    fmt=".1f",
    cmap="coolwarm",
    vmin=0,
    vmax=100,
    ax=axes[0]
)

axes[0].set_title(
    "Category Win Probability (%)"
)

axes[0].set_xlabel("")
axes[0].set_ylabel("")


sns.heatmap(
    df_score,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=0,
    vmax=max(score_probs),
    ax=axes[1]
)

axes[1].set_title(
    "Score Distribution (%)"
)

axes[1].set_xlabel(
    "Win : Loss"
)

axes[1].set_ylabel("")
axes[1].set_yticks([])


plt.tight_layout()

plt.show()
```

```python
def compare_add_drop_scenario(
    baseline_roster,
    opp_roster,
    schedule_df,
    start_date,
    end_date,
    current_my,
    current_opp,
    add_player,
    drop_player,
    season="2025-26",
    n_sim=3000,
    random_seed=12345
):

    # ------------------------------------------
    # 建立 Add / Drop roster
    # ------------------------------------------

    scenario_roster = create_add_drop_roster(
        baseline_roster=baseline_roster,
        drop_player=drop_player,
        add_player=add_player,
        season=season
    )

    # ------------------------------------------
    # Baseline
    # ------------------------------------------

    print("\nRunning BASELINE...")

    baseline = simulate_matchup(
        my_roster=baseline_roster,
        opp_roster=opp_roster,
        schedule_df=schedule_df,
        start_date=start_date,
        end_date=end_date,
        current_my=current_my,
        current_opp=current_opp,
        n_sim=n_sim,
        random_seed=random_seed
    )

    # ------------------------------------------
    # Add / Drop scenario
    # ------------------------------------------

    print("Running ADD / DROP scenario...")

    scenario = simulate_matchup(
        my_roster=scenario_roster,
        opp_roster=opp_roster,
        schedule_df=schedule_df,
        start_date=start_date,
        end_date=end_date,
        current_my=current_my,
        current_opp=current_opp,
        n_sim=n_sim,
        random_seed=random_seed
    )

    # ------------------------------------------
    # Category comparison
    # ------------------------------------------

    categories = [
        "FG%",
        "FT%",
        "Pts",
        "Reb",
        "Ast",
        "Stl",
        "Blk",
        "3PM",
        "TO"
    ]

    rows = []

    for cat in categories:

        baseline_prob = baseline["categories"][cat]["win_prob"]
        scenario_prob = scenario["categories"][cat]["win_prob"]

        rows.append({
            "Category": cat,
            "Baseline": baseline_prob,
            "AddDrop": scenario_prob,
            "Delta": scenario_prob - baseline_prob
        })

    comparison_df = pd.DataFrame(rows)

    # ------------------------------------------
    # Overall comparison
    # ------------------------------------------

    overall = pd.DataFrame({
        "Target": [
            "P(>=5)",
            "P(>=6)",
            "P(>=7)"
        ],
        "Baseline": [
            baseline["p_ge_5"],
            baseline["p_ge_6"],
            baseline["p_ge_7"]
        ],
        "AddDrop": [
            scenario["p_ge_5"],
            scenario["p_ge_6"],
            scenario["p_ge_7"]
        ]
    })

    overall["Delta"] = (
        overall["AddDrop"]
        - overall["Baseline"]
    )

    # ------------------------------------------
    # Print
    # ------------------------------------------

    print("\n" + "=" * 70)
    print("ADD / DROP SCENARIO ANALYSIS")
    print("=" * 70)

    print(f"ADD  : {add_player}")
    print(f"DROP : {drop_player}")

    print("\nCATEGORY WIN PROBABILITY (%)")
    print("-" * 70)

    display_category = comparison_df.copy()

    for col in ["Baseline", "AddDrop", "Delta"]:
        display_category[col] = (
            display_category[col] * 100
        ).round(1)

    print(
        display_category.to_string(index=False)
    )

    print("\nMATCHUP TARGET PROBABILITY (%)")
    print("-" * 70)

    display_overall = overall.copy()

    for col in ["Baseline", "AddDrop", "Delta"]:
        display_overall[col] = (
            display_overall[col] * 100
        ).round(1)

    print(
        display_overall.to_string(index=False)
    )

    # ------------------------------------------
    # Return
    # ------------------------------------------

    return {
        "baseline": baseline,
        "scenario": scenario,
        "category_comparison": comparison_df,
        "overall_comparison": overall,
        "scenario_roster":scenario_roster
    }
```
